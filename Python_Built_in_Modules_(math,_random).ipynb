```python
# 1. Square Root
# Import the math module and find the square root of 144.

# Expected Output:
# 12.0

from math import sqrt

print (sqrt(144))
```

```python
# 2. Power
# Import the math module and find the value of 5 raised to the power of 3.

# Expected Output:
# 125.0

import math

print(math.pow(5,3))
```

```python
# 3. Ceiling and Floor
# Import the math module.
# Find the ceiling and floor values of 7.6.

# Expected Output:
# Ceiling: 8
# Floor: 7

import math
print ("Ceiling:", math.ceil(7.6))
print ("Floor: " , math.floor(7.6))
```

```python
# 4. Factorial
# Import the math module.
# Find the factorial of 6.

# Expected Output:
# 720

import math
print (math.factorial(6))
```

```python
# 5. Greatest Common Divisor (GCD)
# Import the math module.
# Find the GCD of 48 and 18.

# Expected Output:
# 6

import math
print (math.gcd(48,18))
```

```python
# 6. Random Integer
# Import the random module.
# Generate a random integer between 1 and 100.

# Expected Output:
# (Any number between 1 and 100)

import random

print (random.randint(1,100))
```

```python
# 7. Random Choice
# Given:
# fruits = ["Apple", "Banana", "Orange", "Mango"]

# Select one random fruit from the list and print it.

# Expected Output:
# (Any one fruit)

import random
fruits = ["Apple", "Banana", "Orange", "Mango"]

print (random.choice(fruits))
```

```python
# 8. Shuffle a List
# Given:
# numbers = [1, 2, 3, 4, 5]

# Shuffle the list randomly and print it.

# Expected Output:
# (Order changes every time)

import random
numbers = [1, 2, 3, 4, 5]

random.shuffle(numbers)

print (numbers)
```

```python
# 9. Random Sample
# Given:
# students = ["Sai", "John", "Alice", "Bob", "David"]

# Select any 2 unique students randomly using sample().

# Expected Output:
# (Any two unique names)

import random

students = ["Sai", "John", "Alice", "Bob", "David"]

print (random.sample(students,2))
```

```python
# 10. Guess the Number
# Generate a random number between 1 and 10.
# Ask the user to enter a number.
# If the user's number matches the generated number,
# print "You Won!"
# Otherwise print "Try Again!"

import random

num = int(input("Enter a number (1-10): "))

result = random.randint(1, 10)

print("Computer generated:", result)

if num == result:
    print("🎉 You Won!")
else:
    print("❌ Try Again!")
```
