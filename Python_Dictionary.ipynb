```python
# 2. Add a New Key
# Given:
# student = {"name": "John", "age": 22}

# Add a new key "city" with the value "Hyderabad" and print the updated dictionary.

student = {"name": "John", "age": 22}
student.update ({"city":"Hyderabad"})

print (student)
```

```python
# 3. Update a Value
# Given:
# student = {"name": "John", "age": 22}

# Update the age to 23 and print the dictionary.

student = {"name": "John", "age": 22}

student["age"]=23

print (student)
```

```python
# 4. Remove a Key
# Given:
# student = {"name": "John", "age": 22, "city": "Delhi"}

# Remove the "city" key and print the dictionary.

student = {"name": "John", "age": 22, "city": "Delhi"}

student.pop("city")

print (student)
```

```python
# 5. Print Keys and Values
# Given:
# student = {"name": "John", "age": 22, "course": "Python"}

# Print:
# - All keys
# - All values
# - All key-value pairs

student = {"name": "John", "age": 22, "course": "Python"}

print (student.keys())
print (student.values())
print (student.items())
```

```python
# 6. Check if a Key Exists
# Given:
# student = {"name": "John", "age": 22}

# Write a program to check whether the key "age" exists in the dictionary.
# If it exists, print "Key exists"; otherwise, print "Key not found".

student = {"name": "John", "age": 22}

if "age" in student :
  print ("key exists")
else :
  print ("key not found")
```

```python
# 7. Count the Number of Items
# Given:
# student = {"name": "John", "age": 22, "city": "Delhi"}

# Write a program to print the number of key-value pairs in the dictionary.

student = {"name": "John", "age": 22, "city": "Delhi"}

print (len(student))
```

```python

# 8. Merge Two Dictionaries
# Given:
# d1 = {"a": 1, "b": 2}
# d2 = {"c": 3, "d": 4}

# Merge the two dictionaries and print the result.

d1 = {"a": 1, "b": 2}
d2 = {"c": 3, "d": 4}

d1.update(d2)

print(d1)
```

```python
# 8. Merge Two Dictionaries
# Given:
# d1 = {"a": 1, "b": 2}
# d2 = {"c": 3, "d": 4}

# Merge the two dictionaries and print the result.

d1 = {"a": 1, "b": 2}
d2 = {"c": 3, "d": 4}

result = d1 | d2

print(result)
```

```python
# 9. Find the Highest Value
# Given:
# marks = {
#     "Math": 85,
#     "Science": 92,
#     "English": 78
# }

# Write a program to print the subject with the highest marks and its value.

marks = {
    "Math": 85,
    "Science": 92,
    "English": 78
}

subject = max(marks, key=marks.get)

print(subject)
print(marks[subject])
```

```python
# 10. Count Character Frequency
# Write a program to count the frequency of each character in a string using a dictionary.

# Example:
# text = "programming"

# Expected Output:
# {'p': 1, 'r': 2, 'o': 1, 'g': 2, 'a': 1, 'm': 2, 'i': 1, 'n': 1}

text = "programming"

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```
