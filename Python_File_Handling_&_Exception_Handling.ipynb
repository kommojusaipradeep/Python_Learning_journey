```python
# 1. Create a file named student.txt and write:
#    Name: Sai
#    Age: 24

file = open("student.txt", "w")

file.write("Name: Sai\n")
file.write("Age: 24")

file.close()
```

```python
# 2. Read the contents of student.txt and print them.
with open("student.txt", "r") as file :
  print (file.read())
```

```python
# 3. Append the following line to student.txt:
#    Course: Python
#    Then read and print the updated file.

file = open("student.txt","a")
file.write("\ncourse : Python")

file.close ()

with open("student.txt", "r") as file :
  print (file.read())
```

```python
# 4. Read a file and print the total number of lines it contains.
file = open("student.txt", "r")

lines = file.readlines()
print(len(lines))

file.close()
```

```python
# 5. Ask the user for their name and age, then save the details to a file named details.txt.
file = open("details.txt", "w")

name = input("Enter Your Name: ")
age = input("Enter Your Age: ")

file.write("Name: " + name + "\n")
file.write("Age: " + age)

file.close()
```

```python
# 6. Ask the user to enter a number. If they enter text instead of a number, display:
#    "Invalid input. Please enter a number."

try:
    data = int(input("Enter a number: "))
    print("You entered:", data)
except ValueError:
    print("Invalid input. Please enter a number.")
```

```python
# 7. Ask the user to enter two numbers and print their division. Handle:
#    - ValueError
#    - ZeroDivisionError

try:
    num_1 = int(input("Enter num1: "))
    num_2 = int(input("Enter num2: "))

    result = num_1 / num_2
    print("Result:", result)

except ValueError:
    print("Invalid input. Please enter a number.")

except ZeroDivisionError:
    print("Zero cannot be used as the divisor.")
```

```python
# 8. Open a file named data.txt. If the file doesn't exist, print:
#    "File not found."

try:
    file = open("data.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("File not found.")
```

```python
# 9. Read a file containing one number per line and print the sum of all the numbers.

file = open("numbers.txt", "r")

total = 0

for line in file:
    total = total + int(line)

print("Total =", total)

file.close()
```

```python
# 10. Write a program that:
#     - Asks the user for their name and age.
#     - Saves them to user.txt.
#     - Reads and prints the contents of the file.
#     - Uses exception handling to handle invalid age input and file-related errors.

try:
    name = input("Enter your name: ")
    age = int(input("Enter your age: "))

    file = open("user.txt", "w")
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}")
    file.close()

    file = open("user.txt", "r")
    print(file.read())
    file.close()

except ValueError:
    print("Invalid age. Please enter a valid number.")

except FileNotFoundError:
    print("File not found.")

except Exception:
    print("An unexpected error occurred.")
```
