```python
# 1. Square a Number
# Write a lambda function to find the square of a number.
# Example:
# Input: 5
# Output: 25

square = lambda x:x*x
print (square(5))
```

```python
# 2. Add Two Numbers
# Write a lambda function to add two numbers.
# Example:
# Input: 10, 20
# Output: 30

add = lambda a,b: a+b
print(add(10,20))
```

```python
# 3. Find the Larger Number
# Write a lambda function to return the larger of two numbers.
# Example:
# Input: 15, 25
# Output: 25

largest = lambda a,b : a if a>b else b
print (largest(15,25))
```

```python
# 4. Check Even or Odd
# Write a lambda function that returns "Even" if a number is even,
# otherwise returns "Odd".
# Example:
# Input: 8
# Output: Even

num = int(input("Enter a number: "))

check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(num))
```

```python
# 5. Find the Length of a String
# Write a lambda function to find the length of a given string.
# Example:
# Input: "Python"
# Output: 6

word = "Python"

result = lambda x: len(x)

print(result(word))
```

```python
# 6. Multiply Three Numbers
# Write a lambda function to multiply three numbers.
# Example:
# Input: 2, 3, 4
# Output: 24

num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))
num3 = int(input("Enter num3: "))

multiply = lambda x, y, z: x * y * z

print(multiply(num1, num2, num3))
```

```python
# 7. Convert a String to Uppercase
# Write a lambda function that converts a string into uppercase.
# Example:
# Input: "data analyst"
# Output: "DATA ANALYST"

word = "data analyst"

capital = lambda x : x.upper()

print (capital(word))
```

```python
# 8. Sort a List of Tuples Using Lambda
# Given:
# students = [("Sai", 85), ("John", 72), ("Alice", 95), ("Bob", 80)]

# Sort the list based on marks in ascending order using lambda.

students = [("Sai", 85), ("John", 72), ("Alice", 95), ("Bob", 80)]

result = sorted(students, key=lambda x: x[1])

print(result)
```

```python
# 9. Use Lambda with filter()
# Given:
#numbers = [10, 15, 20, 25, 30, 35, 40]

# Use lambda with filter() to print only even numbers.

numbers = [10, 15, 20, 25, 30, 35, 40]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
```

```python
# 10. Use Lambda with map()
# Given:
#numbers = [1, 2, 3, 4, 5]

# Use lambda with map() to create a new list containing
# the square of each number.

numbers = [1, 2, 3, 4, 5]
result = list (map(lambda x:x*x, numbers))

print (result)
```
