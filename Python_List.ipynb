```python
# 5. Find and print the length of a list using a built-in function.
fruits = ["mango","apple","banana","grapes","kiwi"]
print (len(fruits))
```

```python
# 6. Find the largest element in a list without using the `max()` function.
num = [6, 4, 10, 2, 5, 1, 9, 7, 8, 3]
num.sort()
print (num[-1])
```

```python
# 6. Find the largest element in a list without using the `max()` function.
num = [6, 4, 10, 2, 5, 1, 9, 7, 8, 3]

largest_num = num[0]

for i in num:
    if i > largest_num:
        largest_num = i

print(largest_num)
```

```python
# 7. Write a program to remove duplicate elements from a list while preserving the original order.

# Example:
# Input:
# [1, 2, 2, 3, 4, 4, 5]

num = [1, 2, 2, 3, 4, 4, 5]

new_list = []

for i in num:
    if i not in new_list:
        new_list.append(i)

print(new_list)
```

```python
# 8. Reverse a list without using the `reverse()` method or slicing (`[::-1]`).
num = [1, 2, 2, 3, 4, 4, 5]

rev_num = []

for i in range(len(num) - 1, -1, -1):
    rev_num.append(num[i])

print(rev_num)
```

```python
# 9. Count and print the number of even and odd numbers in a list.

# Example:
# Input:
# [10, 15, 21, 34, 50]

# Output:
# Even = 3
# Odd = 2

num = [10, 15, 21, 34, 50]
even = []
odd = []
for i in num:
  if i % 2 == 0:
    even.append (i)
  else :
    odd.append(i)
print (f"even =",len(even))
print (f"odd =",len(odd))
```

```python
# 10. Merge two lists into one list without using the `+` operator.
num1 = [2,5,4,6,3,6,4]
num2 = [5,74,6,8,4,3,6]

num1.extend(num2)
print (num1)
```
