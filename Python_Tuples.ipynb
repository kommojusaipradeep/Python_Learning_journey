```python
# 1.Create a tuple of five fruits.
fruits = ("apple", "mango", "banana", "grapes","kiwi")
print (fruits)
```

```python
# 2.Print the first and last elements.
fruits = ("apple", "mango", "banana", "grapes","kiwi")
print (fruits [0])
print (fruits[-1])
```

```python
# 3.Count how many times 7 appears in a tuple.
num = (2, 5, 7, 7, 8, 3, 4, 7, 5, 3, 1, 7, 7, 7)

count = 0

for i in num:
    if i == 7:
        count += 1

print("7 appears", count, "times")
```

```python
# 4.Find the index of "Python" in a tuple.
languages = ("Java", "C", "Python", "JavaScript", "C++")

index = languages.index("Python")

print(index)
```

```python
# 5.Concatenate two tuples.
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

result = tuple1 + tuple2

print(result)
```

```python
# 6.Convert a list into a tuple.
num = [2, 5, 7, 7, 8, 3, 4, 7, 5, 3, 1, 7, 7, 7]
result = tuple(num)
print (result)
```

```python
# 7.Convert a tuple into a list.
num = (2, 5, 7, 7, 8, 3, 4, 7, 5, 3, 1, 7, 7, 7)
result = list(num)
print (result )
```

```python
# 8.Unpack a tuple into three variables.
student = ("kumar", 25, "Karnataka")

name, age, city = student

print(name)
print(age)
print(city)
```

```python
# 9.Find the maximum and minimum values.
num = (2, 5, 7, 7, 8, 3, 4, 7, 5, 3, 1, 7, 7, 7)

maximum = max(num)
minimum = min(num)

print("Maximum value:", maximum)
print("Minimum value:", minimum)
```

```python
# 10.Check whether an element exists in a tuple

fruits = ("apple", "mango", "banana", "grapes", "kiwi")

if "banana" in fruits:
    print("Element found")
else:
    print("Element not found")
```
