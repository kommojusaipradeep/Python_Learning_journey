```python
#  1. Create a Set
#  Create a set containing the numbers:
#  10, 20, 30, 40, 50
#  Print the set.

num = {10,20,30,40,50}

print ( sorted (num))
```

```python
#  2. Add an Element
#  Create a set:
#  {5, 10, 15}
#  Add 20 to the set and print it.
num = {5, 10, 15}
num.update({20})
print (num)
```

```python
#  3. Remove an Element
#  Create a set:
#  {2, 4, 6, 8, 10}
#  Remove 6 from the set and print the updated set.

num = {2, 4, 6, 8, 10}

num.remove(6)

print(num)
```

```python
#  4. Check Membership
#  Create a set:
#  {"apple", "banana", "orange"}
#  Check whether "banana" exists in the set.

#  Expected Output:
#  banana is present
#  or
#  banana is not present

fruits = {"apple", "banana", "orange"}

if "banana" in fruits:
  print ("banana is present")
else :
  print ("banana is not present")
```

```python
#  5. Find the Length
#  Create a set:
#  {11, 22, 33, 44, 55, 66}
#  Print the total number of elements in the set.

num = {11, 22, 33, 44, 55, 66}

print (len(num))
```

```python
#  6. Union of Two Sets
#  Given:
# A = {1, 2, 3, 4}
# B = {3, 4, 5, 6}

# Find the union of the two sets.

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.union(B)

print (result)
```

```python
# # 7. Intersection of Two Sets
# # Given:
# A = {10, 20, 30, 40}
# B = {30, 40, 50, 60}

# # Find the common elements.

A = {10, 20, 30, 40}
B = {30, 40, 50, 60}

result = A.intersection(B)

print (result)
```

```python
#  8. Difference Between Sets
#  Given:
# A = {1, 2, 3, 4, 5}
# B = {3, 4, 5, 6, 7}

# Find the elements that are present only in A.

A = {1, 2, 3, 4, 5}
B = {3, 4, 5, 6, 7}

result = A-B
print (result)
```

```python
# 9. Remove Duplicates from a List
# Given:
# numbers = [1, 2, 2, 3, 4, 4, 5, 6, 6]

# Convert the list into a set to remove duplicate values and print the result.

numbers = [1, 2, 2, 3, 4, 4, 5, 6, 6]

result = set(numbers)
print (result)
```

```python
# 10. Find Unique Characters
# Write a program that takes a string from the user and prints all the unique characters using a set.

# Example:
# Input:
# programming

# Expected Output:
# {'p', 'r', 'o', 'g', 'a', 'm', 'i', 'n'}

word = input("Enter a Word")

result = set(word)

print (result)
```
